### Activity: write/decode a message in binary ascii code

```python
# Use the code below to check your message

def binary_to_text(binary_string):
    """
    Converts a binary ASCII string to English text.

    Parameters:
    binary_string (str): A string of binary numbers separated by spaces.

    Returns:
    str: The decoded English text.
    """
    # Split the input into individual binary numbers
    binary_values = binary_string.split()

    # Convert each binary value to its ASCII character
    ascii_characters = [chr(int(b, 2)) for b in binary_values]

    # Join all characters into a single string
    return ''.join(ascii_characters)

# Example
binary_input = "01001000 01100101 01101100 01101100 01101111"
text_output = binary_to_text(binary_input)
print(text_output)  # Output: Hello
```

###### Slicing ?
- You can return a range of characters by using the slice syntax
- Use indexing to specify the part of the string you want to output

```python
string = "Hello World"

# Print characters from position 0 up to (but not including) position 5
# This will print "Hello"
print(string[0:5])

# Same as above – if you leave out the start index, Python assumes it starts at 0
print(string[:5])  # Also prints "Hello"

# Print characters from position 6 to the end of the string
# This will print "World"
print(string[6:])

# Print the last 5 characters of the string
# Negative indexes count from the end: -5 means 5 characters from the end
print(string[-5:])  # Prints "World"
```

###### Concatenation
- You can use `+` to combine two strings

```python
a = "Hello"
b = "World"

# Combine (or "concatenate") the strings 'a' and 'b' with a space in between
# " " is a string that contains just one space character
c = a + " " + b
d = a + b

# Print the final combined message: "Hello World"
print(c)
print(d)
```

##### Arthmetic Operators
- addition: `+`
- substraction: `-`
- multiplicaiton: `*`
- division: `/`
- modules: `%`
  - outputs remainder
- power: `**`
- integer division: `//`
  - discard remainder and outputs integer part of the quotient

```python

integer_1 = 7
integer_2 = 2

# Add the two numbers
print(integer_1 + integer_2)  # Output: 9

# Subtract the second number from the first
print(integer_1 - integer_2)  # Output: 5

# Multiply the two numbers
print(integer_1 * integer_2)  # Output: 14

# Divide the first number by the second (always gives a float)
print(integer_1 / integer_2)  # Output: 3.5

# Modulus: gives the remainder of the division
print(integer_1 % integer_2)  # Output: 1 (because 7 ÷ 2 = 3 with a remainder of 1)

# Exponentiation: raises the first number to the power of the second
print(integer_1 ** integer_2)  # Output: 49 (because 7^2 = 49)

# Int division: divides and rounds down to the nearest whole number
print(integer_1 // integer_2)  # Output: 3

# Use an f-string to show the relationship: a = b * (a // b) + (a % b)
print(f"{integer_1} = {integer_2} * {integer_1 // integer_2} + {integer_1 % integer_2}")
# Output: 7 = 2 * 3 + 1
```

#### Boolean
- Boolean represents `True` and `False` in python.
- Booleans are often the result of comparisons or logical operations (e.g., 5 > 3, x == y, not True).
- They are commonly used in conditional statements (e.g., if, while) to control the flow of a program based on whether a condition is true or false.

```python
bool1 = True
bool2 = False

# Print the value of bool1 (will print: True)
print(bool1)

# Print the value of bool2 (will print: False)
print(bool2)
```

##### Logical Operators  
Operators used to combine conditional statements  

- `and` : Returns `True` if both statements are true  
- `or` : Returns `True` if at least one statement is true  
- `not` : Reverses the result, returns `False` if the result is true  


```python
# Use the 'and' operator: this returns True only if BOTH sides are True
# True and False → False (because both are not true)
bool6 = True and False

# Use the 'or' operator: this returns True if AT LEAST ONE side is True
# True or False → True (because one side is true)
bool7 = True or False

# Use the 'not' operator: this reverses the Boolean value
# not True → False
bool8 = not True

# Print the results
print(bool6)  # False
print(bool7)  # True
print(bool8)  # False
```

##### Comparison Operators  
Operators used to compare two values  

- `==` : Equal  
- `!=` : Not equal  
- `>` : Greater than  
- `<` : Less than  
- `>=` : Greater than or equal to  
- `<=` : Less than or equal to  


```python
# Check if 5 is greater than 3
# This is True because 5 is bigger than 3
bool3 = 5 > 3

# Check if 5 is equal to 3
# This is False because 5 is not the same as 3
bool4 = 5 == 3

# Check if 5 is NOT equal to 3
# This is True because 5 is different from 3
bool5 = 5 != 3

# Print the result of each comparison
print(bool3)  # True
print(bool4)  # False
print(bool5)  # True
```

#### Conditionals
- if condition: code runs if condition is True
- elif another_condition: code runs if this new condition is True

###### [List methods](https://www.w3schools.com/python/python_ref_list.asp):
- `.append( )` : adds an element to end of list
- `.insert()`: add an element to a specified index
-`.pop()`: removes element at the specified position
- `len()`:finds the length of a list (also works for strings)
-`index()`:finds the index of the first element with the specified value (also works for strings)

```python
my_list = [1, 2, 3, 4, 5]  # Start with a list of numbers

# Add the number 6 to the end of the list
my_list.append(6)
print(my_list)  # Output: [1, 2, 3, 4, 5, 6]

# Insert the number 0 at position 0 (the beginning of the list)
# .insert(value, position)
my_list.insert(0, 0)
print(my_list)  # Output: [0, 1, 2, 3, 4, 5, 6]

# Remove the item at index 0 (which is 0)
my_list.pop(0)
print(my_list)  # Output: [1, 2, 3, 4, 5, 6]

# Print how many items are in the list
print(len(my_list))  # Output: 6

# Find the index (position) of the number 3 in the list
print(my_list.index(3))  # Output: 2 (because 3 is at position 2)

```

Method 1 for creating a dictionary: Create a Dictionary All at Once

###### Common Dictionary Methods and Operations:
- `.get():` get value stored at a certain key
- `.pop()`: output the pair and remove pair from list  
- `.keys()`: get all keys
- `.values()`: get all values

```python
# Start with an empty dictionary called 'planet'
planet = {}

# Add some key–value pairs to the dictionary
planet["moon"] = 2              # Earth has 2 moons (in this example)
planet["galaxy"] = "milkyway"   # Earth is in the Milky Way galaxy
planet["name"] = "earth"        # The name of the planet is Earth

# Get the value associated with the key "moon"
print(planet.get("moon"))       # Output: 2

# Remove the key "galaxy" and return its value
print(planet.pop("galaxy"))     # Output: "milkyway"

# Show all the keys in the dictionary
print(planet.keys())            # Output: dict_keys(['moon', 'name'])

# Show all the values in the dictionary
print(planet.values())          # Output: dict_values([2, 'earth'])

```

## Iteration

- in programming, we use iteration to go through:

  - strings

### For Loop
- use to loop through a sequence (linear data structures or a string)


```python
# it can be used to print the content in a list
fruits = ["apple", "banana", "cherry"]
for fruit in fruits: # Loop through each item in the list
    print(fruit) # Print the current fruit
```

#### Using `range()`
- used when you want to repeat something a number of times


```python
# basic form: range(stop)
for i in range(5): # Loop from 0 to 4 (5 is not included)
    print(i) # Print the current number

#full form: range(start, stop, step)
for i in range(2, 10, 2):  # Start at 2, go up to (but not including) 10, step by 2
    print(i) # Print the current number

# you can also use this to loop through a linear data structure:
fruits = ["apple", "banana", "cherry"]
for i in range(len(fruits)): # Loop through the indices of the list
    print(f"fruit {i} is {fruits[i]}") # Print the current fruit
print(" ")
# you can use the full form to loop through a linear data structure as well:
animal = ["dog", "cat", "bird","lizard", "bugs","mouse","horse"]
for i in range(1,len(animal),2): # Loop through the indices of the list
    print(f"animal {i} is {animal[i]}") # Print the current fruit
```

### While Loops
- Use while when you don’t know how many times to repeat ahead of time — just repeat while a condition is true

```python
count = 0  # Start count at 0

while count < 3:              # Repeat the loop while count is less than 3
    print("Repeating...")     # Print a message
    count += 1                # Add 1 to count each time the loop runs
```

### Write to a File
- to save data, we can use python to create and write to a file
- useful in scientific context
  - eg.  saving experiment results, simulation outputs, or data logs
  

```python
# Let's say we're logging hourly temperature readings in a lab
temperatures = [21.3, 21.6, 22.0, 22.5, 22.1]

# Open a file in write mode ('w') – this will create or overwrite the file
with open("temperature_log.txt", "w") as file:
    for temp in temperatures:
        file.write(f"{temp}\n")  # Write each temperature on its own line
```

- `\n` creates a new line for each data point.
- The `with` statement automatically closes the file when you're done.

### Reading from a File
- you can output a recorded file in python too


```python
# Open the file in read mode ('r')
with open("temperature_log.txt", "r") as file:
    for line in file:
        temp = float(line.strip())  # Remove newline and convert to number
        print(f"Read temp: {temp}°C")
```
