### Step 11: Create a function to describe your galaxy

Functions are reusable blocks of code. Define a function using `def`, give it a name, and place the code inside the indented block.
Functions can also use **parameters** to receive information and `return` to send a result back.

Create a function called `describe_planet` with parameters for a planet's name and temperature. Return a sentence describing the planet, then print the returned result.

Try calling the function with different planets and values.

```python
# Define a function that describes a planet
def describe_planet(planet_name, temperature):
    """Return a sentence describing the planet."""
    return f"{planet_name.title()} has a temperature of {temperature}°C."


# Call the function using existing planets
print(describe_planet(pluto["name"], pluto["temp_in_c"]))
print(describe_planet(earth["name"], earth["temp_in_c"]))

# Try the function with another planet
print(describe_planet("Mars", -63))
```
