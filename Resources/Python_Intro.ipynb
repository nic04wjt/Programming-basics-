### Step 10: Use input + a while loop to add new stars to your galaxy

```python
'''
input(): returns a string
name=input("what is your name?")
print( f"hello {name}")
'''
new_star1 = input("Name a new star: ")
stars.append(new_star1)
want_to_add_new_star=input("Do you want to add a new star to your galaxy? (yes/no): ")

while want_to_add_new_star == "yes":
  new_star=input("Name a new star: ")
  stars.append(new_star)
  want_to_add_new_star=input("Do you want to add a new star to your galaxy? (yes/no): ")
```

### Step 11: Save your list of stars as a file

```python
# create file:
# f=open("stars.txt","x") x=create file
# f.close to close file
# to write information into file:
with open ("stars.txt","w") as f: # with: closes file after indented block
  for star in stars:
    f.writelines(f"{star} ")
#to read a file:
with open("stars.txt","r") as f:
  my_star=f.readlines()
  print(my_star)
```
